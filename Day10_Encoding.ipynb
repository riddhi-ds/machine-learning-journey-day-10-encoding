***Day 10 - Encoding***


***Encoding*** is the process of converting categorical data into numerical form so that machine learning algorithms can understand and process it.


Why do we need Encoding ?

Suppose our dataset has:

Gender

Male

Female

Male

Female


*Male and Female are categorical data but ML algo works with numerical data* 


So we need to convert Male / Female -> Numerical values

For example :

Male -> 1

Female -> 0


Main types we'll learn today is :

a. ***Label Encoding***

b. ***One - Hot Encoding***

c. ***Ordinal Encoding***


```python
import pandas as pd

titanic = pd.read_csv("titanic.csv")

titanic.head()
```

```python
titanic.columns
```

```python
titanic['Sex'].unique
```

***Label Encoding***

**Label Encoding** is generally suitable for target columns , *especially when the categories are just class names*

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

titanic['Sex_encoded'] = le.fit_transform(titanic['Sex'])

titanic[['Sex','Sex_encoded']].head()
```

***One - Hot Encoding***

**One - Hot Encoding** is a process of converting categorical data into numerical data by *creating seperate binary (1/0) column for each category*  

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output = False,drop = 'first')

encoded = ohe.fit_transform(titanic[['Embarked']])

print(encoded)
```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(['Embarked'])
).astype(int)

print(encoded_df)
```

***Ordinal Encoding***

**Ordinal Encoding** is a process of converting catgeorial data to numerical data while *preserving their natural order or ranking*

```python
titanic['Education'] = (['School','HSC','Bachelor','Master','PHD'] * 105)[:418] 
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['School','HSC','Bachelor','Master','PHD']])

titanic['Education_encoded'] = oe.fit_transform(titanic[['Education']]).astype(int)

print(titanic[['Education','Education_encoded']].head(10))
```
